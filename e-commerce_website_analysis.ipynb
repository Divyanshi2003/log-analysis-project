```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("E-commerce Website Logs.csv")
```

```python
df.head()
```

```python
df['gender'].unique()
```

```python
df['gender_short']=df['gender'].map({'Male': 'M','Female': 'F','Unknown': None})
df = df.drop('gender', axis=1)
```

```python
df.head()
```

```python
ip_frequency = df['ip'].value_counts()
print(ip_frequency)
```

```python
medium_count = df['accessed_Ffom'].value_counts()
print(medium_count)
```

```python
df['accessed_date_datetime'] = pd.to_datetime(df['accessed_date'])
df['date']= df['accessed_date_datetime'].dt.date
df
```

```python
customer_daily_avg = df.groupby(['ip','date'])['duration_(secs)'].mean().reset_index()
customer_daily_avg
 
```

```python
df
```

```python
average_duration_overall = df['duration_(secs)'].mean()
print("Average duration overall: in (s)", average_duration_overall)
```

```python
avg_by_age = df.groupby('age')['duration_(secs)'].mean().reset_index()

avg_by_age
```

```python
# Create a barplot
plt.figure(figsize=(15,6))
sns.barplot(data=avg_by_age, x='age', y='duration_(secs)', palette='viridis')

plt.title('Average Session Duration by Age')
plt.xlabel('Age')
plt.ylabel('Average Duration (secs)')
plt.show()
```

```python
df['hour']=df['accessed_date_datetime'].dt.hour     # 0 to 23
df
```

```python
peak_hours = df['hour'].value_counts().sort_index()
peak_hours = peak_hours.to_frame().reset_index()
peak_hours.columns = ['hour', 'session_count']
```

```python
# Create a barplot
plt.figure(figsize=(15,6))
sns.barplot(data=peak_hours, x='hour', y='session_count', palette='viridis')

plt.title('peak usage hours')
plt.xlabel('hour')
plt.ylabel('session count')
plt.show()
```

```python
df['returned_bool'] = df['returned'].apply(lambda x: True if x == 'Yes' else False)

```

```python
orders_returned_array=df['returned_bool'].value_counts()
print(orders_returned_array)
```

```python
#percent of orders returned
percent_returned = (orders_returned_array[True] / orders_returned_array.sum()) * 100
print(f"Percentage of orders returned: {percent_returned:.2f}%")
```

```python
df
```

```python
returned_amount=df.groupby('returned_bool')['returned_amount'].sum().reset_index()
```

```python
customers_by_language=df.groupby('returned_bool')['language'].value_counts()
```

```python
total_customers_by_language = df['language'].value_counts().reset_index()
total_customers_by_language
```

```python
#customers that returned items grouped by language
customers_by_language.loc[True]
```

```python
merged = pd.merge(customers_by_language.loc[True]
, total_customers_by_language, on='language', suffixes=('churn', 'total'))
```

```python
merged
```

```python
merged['return_rate'] =( merged['countchurn'] / merged['counttotal'] * 100).round(1)
```

```python
merged
```

```python
# Create a barplot
plt.figure(figsize=(25,6))
sns.barplot(data=merged, x='language', y='return_rate', palette='viridis')

plt.title('return rate by language')
plt.xlabel('languages')
plt.ylabel('return rate (%)')
plt.show()
```
